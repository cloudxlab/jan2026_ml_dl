Each is having an accuracy of 90% for cancer prediction.

We create a team having all three. Whenever we need to predict, we ask all of them. and We take the majority answer.

What will be the accuracy of the team?

A. Greater than 90%
B. 90%
C. Less than 90%

Compute the exact accuracy.

```python

# when will they be wrong - two or more are wrong
1 - (.1 * .1 * .9 * 3 + .1 * .1 * .1)
```

```python
def compute_accuracy_hard_voting3(a1, a2, a3):
    # return accuracy of ensemble
    p1 = a1 * a2 * a3  
    p2 = a1 * a2 * (1-a3) + a1 * (1-a2) * a3 + (1-a1) * a2 * a3
    return p1 + p2 
```
