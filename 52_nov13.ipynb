```python
import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers

np.random.seed(42)

n = 5000

area = np.random.uniform(0.5, 3, n)
bedrooms = np.random.randint(1, 6, n)
age = np.random.uniform(0, 1, n)

X = np.column_stack([area, bedrooms, age])

y = (
    100 * area
    + 20 * bedrooms
    - 30 * age
    + 15 * area**2
    + 10 * area * bedrooms
    + np.random.normal(0, 5, n)
)
```

```python
inputs = keras.Input(shape=(3,))

# Deep path
hidden1 = layers.Dense(32, activation="relu")(inputs)
hidden2 = layers.Dense(16, activation="relu")(hidden1)
hidden3 = layers.Dense(8, activation="relu")(hidden2)

# Combine wide and deep paths
combined = layers.Concatenate()([inputs, hidden3])

# Output layer
output = layers.Dense(1)(combined)

model = keras.Model(inputs=inputs, outputs=output)

model.compile(
    optimizer="adam",
    loss="mse",
    metrics=["mae"]
)

model.summary()
```

```python
history = model.fit(
    X, y,
    epochs=100,
    batch_size=32,
    validation_split=0.2
)
```
