```python
# We want to find out the values of w that really fit our data

def error(X, y_actual, w):
  # shape of X : m x n
  # shape of y : m
  # shape of w = n+1

  y_predicted = model_predict(X, w)
  mse = np.sum((y_predicted - y_actual)**2)
  return mse
```

```python
import numpy as np

X = np.array([
    [1],
    [2],
    [5]
])

y = np.array([5.8,6.5,7.5])

import matplotlib.pyplot as plt
# Create scatter plot
plt.scatter(X.reshape(-1), y)
```

```python

def loss(X, y, w):
    """
    Mean Squared Error loss.

    Parameters:
        X : (n, d) numpy array
        y : (n,) numpy array
        w : (d,) or (d,1) numpy array

    Returns:
        float: Mean Squared Error
    """
    y_pred = X.dot(w[1:]) + w[0]
    # y_pred = y_pred.ravel()      # Ensure shape is (n,)
    return np.mean((y - y_pred) ** 2)

# Example
w = np.array([1.5, 2])

print(loss(X, y, w))
```

```python
import numpy as np
import matplotlib.pyplot as plt

# # Data
# X = np.array([1, 2, 5])
# y = np.array([5.8, 6.5, 7.5])

# # Line parameters
# m = 0.42
# c = 5.1

def plot(X, y, m, c):
    # Scatter plot
    plt.scatter(X, y, color='blue', label='Data')

    # Generate points for the line
    x_line = np.linspace(X.min(), X.max(), 100)
    y_line = m * x_line + c

    # Plot the line
    plt.plot(x_line, y_line, color='red', label=f'y = {m:.2f}x + {c:.2f}')

    # Labels and title
    plt.xlabel("X")
    plt.ylabel("y")
    plt.title("Scatter Plot with Best Fit Line")
    plt.legend()
    plt.grid(True)

    plt.show()

plot(X.reshape(-1), y, 2, 1.5)
```

```python
def gradients(loss, X, y, w):
     d = 0.00000001
     base_error = loss(X, y, w)
     print("base_error: ", base_error)
     diff1 = (loss(X, y, w + np.array([d, 0])) - base_error) / d # return on investment
     diff2 = (loss(X, y, w + + np.array([0, d])) - base_error) / d # return on investment
     return np.array([diff1, diff2])
```

```python
w = [0, 1.5]
for i in range(5000):
    base_error = loss(X, y, w)
    eta = 0.01 # learning rate
    grads = gradients(loss, X, y, w)
    w = w -  grads * eta # gradient descent step. update the weights
    base_error = loss(X, y, w)
    print("base_error: ", base_error)
```
