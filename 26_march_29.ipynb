```python
# Utilizing the Matrix Multiplication for calculating the cosine similarities between people
```

```python
import numpy as np
M = np.array([ # Three people and four movies rating 
    [.1, .2, .4],
    [.2, .2, .3],
    [.3, .2, .1],
    [.1, .2, .1],
])
```

```python
sqM = np.square(M)
print(sqM)
norms = np.sqrt(sqM.sum(axis=0))
norms
```

```python
uM = M / norms
```

```python
uM
```

```python
umT = uM.T
umT
```

```python
umT @ uM
```
