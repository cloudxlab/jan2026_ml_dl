Q: You have N people in a a room, write a python function that takes n and gives out the prob  that at least 2 people share the birthday.
A: 

````python
def prob_sharing_birthday(n):
    if n >= 365:
        return 1
    
    num = 365
    p = 1
    while n > 0:
        p *= num / 365
        num -= 1
        n -= 1
    return 1 - p
````
