```python
def huber_loss(y_true, y_pred, delta):
    total = 0
    for y, yp in zip(y_true, y_pred):
        err = abs(y - yp)
        if err <= delta:
            total += 0.5 * (y - yp) ** 2
        else:
            total += delta * err - 0.5 * delta ** 2
    return total / len(y_true)

print(huber_loss([5, 2, 7], [4.8, 2.5, 10], delta=1))  # ≈ 0.8817
print(huber_loss([1, 2, 3], [1, 2, 3], delta=1))        # 0.0
```

```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression, Lasso
from sklearn.pipeline import make_pipeline
from sklearn.metrics import mean_squared_error


# =========================================================
# 1. Create a noisy dataset
# =========================================================

np.random.seed(42)

X = np.linspace(-3, 3, 30).reshape(-1, 1)

# Underlying relationship
y = 5 * X.ravel()**2 + 2* X.ravel()**3 #+ np.sin(X.ravel())

# Add noise
y += np.random.normal(0, 0.5, size=len(X))


# =========================================================
# 2. Train / test split
# =========================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.4,
    random_state=42
)


# =========================================================
# 3. Use a high-degree polynomial
# =========================================================

degree = 15


# ---------------------------------------------------------
# Without regularization
# ---------------------------------------------------------

linear_model = make_pipeline(
    PolynomialFeatures(degree=degree),
    LinearRegression() #  Loss = (y - yp)**2 
)


# ---------------------------------------------------------
# L1 regularization
# ---------------------------------------------------------

lasso_model = make_pipeline(
    PolynomialFeatures(degree=degree),
    Lasso(alpha=0.01, max_iter=100000) # Loss = (y - yp)**2 + 0.01 * sum(abs(weights)
)


# Train both models
linear_model.fit(X_train, y_train)
lasso_model.fit(X_train, y_train)


# =========================================================
# 4. Predictions
# =========================================================

y_train_linear = linear_model.predict(X_train)
y_test_linear = linear_model.predict(X_test)

y_train_lasso = lasso_model.predict(X_train)
y_test_lasso = lasso_model.predict(X_test)


# =========================================================
# 5. Compare MSE
# =========================================================

print("=" * 70)
print("MODEL PERFORMANCE")
print("=" * 70)

print("\nPolynomial Regression")
print(f"Training MSE : {mean_squared_error(y_train, y_train_linear):.6f}")
print(f"Test MSE     : {mean_squared_error(y_test, y_test_linear):.6f}")

print("\nPolynomial Regression + L1 (Lasso)")
print(f"Training MSE : {mean_squared_error(y_train, y_train_lasso):.6f}")
print(f"Test MSE     : {mean_squared_error(y_test, y_test_lasso):.6f}")


# =========================================================
# 6. Print parameters
# =========================================================

poly_linear = linear_model.named_steps["polynomialfeatures"]
reg_linear = linear_model.named_steps["linearregression"]

poly_lasso = lasso_model.named_steps["polynomialfeatures"]
reg_lasso = lasso_model.named_steps["lasso"]

feature_names = poly_linear.get_feature_names_out(["x"])


print("\n" + "=" * 70)
print("PARAMETERS WITHOUT REGULARIZATION")
print("=" * 70)

print(f"Intercept: {reg_linear.intercept_:.6f}")

for feature, coefficient in zip(
    feature_names,
    reg_linear.coef_
):
    print(f"{feature:>5s} : {coefficient:12.6f}")


print("\n" + "=" * 70)
print("PARAMETERS WITH L1 REGULARIZATION")
print("=" * 70)

print(f"Intercept: {reg_lasso.intercept_:.6f}")

for feature, coefficient in zip(
    feature_names,
    reg_lasso.coef_
):
    print(f"{feature:>5s} : {coefficient:12.6f}")


# =========================================================
# 7. Plot both fitted curves
# =========================================================

x_plot = np.linspace(-3, 3, 500).reshape(-1, 1)

y_linear_plot = linear_model.predict(x_plot)
y_lasso_plot = lasso_model.predict(x_plot)


plt.figure(figsize=(12, 7))

# Training data
plt.scatter(
    X_train,
    y_train,
    s=60,
    label="Training data"
)

# Test data
plt.scatter(
    X_test,
    y_test,
    s=60,
    marker="x",
    label="Test data"
)

# Polynomial regression
plt.plot(
    x_plot,
    y_linear_plot,
    linewidth=2,
    label="Degree 15 - No regularization"
)

# Lasso
plt.plot(
    x_plot,
    y_lasso_plot,
    linewidth=3,
    label="Degree 15 - L1 regularization"
)

plt.xlabel("x")
plt.ylabel("y")
plt.title("Effect of L1 Regularization on Polynomial Regression")
plt.legend()
plt.grid()

plt.ylim(-5, 10)

plt.show()


# =========================================================
# 8. Compare coefficient magnitudes
# =========================================================

plt.figure(figsize=(12, 6))

x_coeff = np.arange(len(feature_names))

plt.plot(
    x_coeff,
    reg_linear.coef_,
    marker="o",
    label="No regularization"
)

plt.plot(
    x_coeff,
    reg_lasso.coef_,
    marker="o",
    label="L1 regularization"
)

plt.xticks(
    x_coeff,
    feature_names,
    rotation=45
)

plt.xlabel("Polynomial feature")
plt.ylabel("Coefficient value")
plt.title("Effect of L1 Regularization on Parameters")
plt.legend()
plt.grid()

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression, Lasso, Ridge
from sklearn.pipeline import make_pipeline
from sklearn.metrics import mean_squared_error


# =========================================================
# 1. Create a noisy dataset
# =========================================================

np.random.seed(42)

X = np.linspace(-3, 3, 30).reshape(-1, 1)

# Underlying relationship
y = 0.5 * X.ravel()**2 + np.sin(X.ravel())

# Add noise
y += np.random.normal(0, 1.0, size=len(X))


# =========================================================
# 2. Train / test split
# =========================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.4,
    random_state=42
)


# =========================================================
# 3. High-degree polynomial models
# =========================================================

degree = 15

# No regularization
linear_model = make_pipeline(
    PolynomialFeatures(degree=degree),
    LinearRegression()
)

# L1 regularization
lasso_model = make_pipeline(
    PolynomialFeatures(degree=degree),
    Lasso(alpha=0.01, max_iter=100000)
)

# L2 regularization
ridge_model = make_pipeline(
    PolynomialFeatures(degree=degree),
    Ridge(alpha=0.01)
)


# =========================================================
# 4. Train all models
# =========================================================

models = {
    "No Regularization": linear_model,
    "L1 (Lasso)": lasso_model,
    "L2 (Ridge)": ridge_model
}

for name, model in models.items():
    model.fit(X_train, y_train)


# =========================================================
# 5. Compare training and test errors
# =========================================================

print("=" * 75)
print("MODEL PERFORMANCE")
print("=" * 75)

for name, model in models.items():

    train_pred = model.predict(X_train)
    test_pred = model.predict(X_test)

    train_mse = mean_squared_error(
        y_train,
        train_pred
    )

    test_mse = mean_squared_error(
        y_test,
        test_pred
    )

    print(f"\n{name}")
    print(f"Training MSE : {train_mse:.6f}")
    print(f"Test MSE     : {test_mse:.6f}")


# =========================================================
# 6. Print parameters
# =========================================================

print("\n" + "=" * 75)
print("MODEL PARAMETERS")
print("=" * 75)

for name, model in models.items():

    print("\n" + "-" * 75)
    print(name)
    print("-" * 75)

    polynomial = model.named_steps["polynomialfeatures"]

    # Get whichever regression model is in the pipeline
    if "linearregression" in model.named_steps:
        regression = model.named_steps["linearregression"]

    elif "lasso" in model.named_steps:
        regression = model.named_steps["lasso"]

    elif "ridge" in model.named_steps:
        regression = model.named_steps["ridge"]

    feature_names = polynomial.get_feature_names_out(["x"])

    print(f"Intercept: {regression.intercept_:.6f}")
    print("\nCoefficients:")

    for feature, coefficient in zip(
        feature_names,
        regression.coef_
    ):
        print(
            f"{feature:>5s} : "
            f"{coefficient:12.6f}"
        )


# =========================================================
# 7. Plot all three fitted curves
# =========================================================

x_plot = np.linspace(-3, 3, 1000).reshape(-1, 1)

plt.figure(figsize=(12, 7))

# Training data
plt.scatter(
    X_train,
    y_train,
    s=60,
    label="Training data"
)

# Test data
plt.scatter(
    X_test,
    y_test,
    s=60,
    marker="x",
    label="Test data"
)


# Plot all models
for name, model in models.items():

    y_plot = model.predict(x_plot)

    plt.plot(
        x_plot,
        y_plot,
        linewidth=2,
        label=name
    )


plt.xlabel("x")
plt.ylabel("y")
plt.title(
    "Degree-15 Polynomial Regression: "
    "No Regularization vs L1 vs L2"
)

plt.legend()
plt.grid()
plt.ylim(-5, 10)

plt.show()


# =========================================================
# 8. Compare coefficients
# =========================================================

polynomial = linear_model.named_steps["polynomialfeatures"]
feature_names = polynomial.get_feature_names_out(["x"])

linear_coef = linear_model.named_steps[
    "linearregression"
].coef_

lasso_coef = lasso_model.named_steps[
    "lasso"
].coef_

ridge_coef = ridge_model.named_steps[
    "ridge"
].coef_


plt.figure(figsize=(14, 7))

x_coeff = np.arange(len(feature_names))

plt.plot(
    x_coeff,
    linear_coef,
    marker="o",
    label="No regularization"
)

plt.plot(
    x_coeff,
    lasso_coef,
    marker="o",
    label="L1 (Lasso)"
)

plt.plot(
    x_coeff,
    ridge_coef,
    marker="o",
    label="L2 (Ridge)"
)

plt.xticks(
    x_coeff,
    feature_names,
    rotation=45
)

plt.xlabel("Polynomial feature")
plt.ylabel("Coefficient value")

plt.title(
    "Effect of L1 and L2 Regularization "
    "on Polynomial Coefficients"
)

plt.legend()
plt.grid()

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression, Lasso, Ridge, ElasticNet
from sklearn.pipeline import make_pipeline
from sklearn.metrics import mean_squared_error


# =========================================================
# 1. Create a noisy dataset
# =========================================================

np.random.seed(42)

X = np.linspace(-3, 3, 30).reshape(-1, 1)

# Underlying relationship
y = 0.5 * X.ravel()**2 + np.sin(X.ravel())

# Add noise
y += np.random.normal(0, 1.0, size=len(X))


# =========================================================
# 2. Train / test split
# =========================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.4,
    random_state=42
)


# =========================================================
# 3. High-degree polynomial models
# =========================================================

degree = 15

models = {

    "No Regularization": make_pipeline(
        PolynomialFeatures(degree=degree),
        LinearRegression()
    ),

    "L1 (Lasso)": make_pipeline(
        PolynomialFeatures(degree=degree),
        Lasso(
            alpha=0.01, # Loss = (y-yp)**2 + 0.01 * sum(abs(weights))
            max_iter=100000
        )
    ),

    "L2 (Ridge)": make_pipeline(
        PolynomialFeatures(degree=degree),
        Ridge(
            alpha=0.01 # # Loss = (y-yp)**2 + 0.01 * sum(weights**2))
        )
    ),

    "Elastic Net": make_pipeline(
        PolynomialFeatures(degree=degree),
        ElasticNet(
            alpha=0.01,
            l1_ratio=0.4,
            max_iter=100000
        ) # Loss = (y-yp)**2 + 0.6 * 0.01 * sum(weights**2)) + 0.4 * 0.01 * sum(abs(weights))
    )
}


# =========================================================
# 4. Train all models
# =========================================================

for name, model in models.items():
    model.fit(X_train, y_train)


# =========================================================
# 5. Compare training and test errors
# =========================================================

print("=" * 80)
print("MODEL PERFORMANCE")
print("=" * 80)

for name, model in models.items():

    train_pred = model.predict(X_train)
    test_pred = model.predict(X_test)

    train_mse = mean_squared_error(
        y_train,
        train_pred
    )

    test_mse = mean_squared_error(
        y_test,
        test_pred
    )

    print(f"\n{name}")
    print(f"Training MSE : {train_mse:.6f}")
    print(f"Test MSE     : {test_mse:.6f}")


# =========================================================
# 6. Print parameters
# =========================================================

print("\n" + "=" * 80)
print("MODEL PARAMETERS")
print("=" * 80)

for name, model in models.items():

    print("\n" + "-" * 80)
    print(name)
    print("-" * 80)

    polynomial = model.named_steps["polynomialfeatures"]

    # Get regression model
    if "linearregression" in model.named_steps:
        regression = model.named_steps["linearregression"]

    elif "lasso" in model.named_steps:
        regression = model.named_steps["lasso"]

    elif "ridge" in model.named_steps:
        regression = model.named_steps["ridge"]

    elif "elasticnet" in model.named_steps:
        regression = model.named_steps["elasticnet"]

    feature_names = polynomial.get_feature_names_out(["x"])

    print(f"Intercept: {regression.intercept_:.6f}")

    print("\nCoefficients:")

    for feature, coefficient in zip(
        feature_names,
        regression.coef_
    ):
        print(
            f"{feature:>5s} : "
            f"{coefficient:12.6f}"
        )


# =========================================================
# 7. Plot all four fitted curves
# =========================================================

x_plot = np.linspace(-3, 3, 500).reshape(-1, 1)

plt.figure(figsize=(13, 8))

# Training data
plt.scatter(
    X_train,
    y_train,
    s=60,
    label="Training data"
)

# Test data
plt.scatter(
    X_test,
    y_test,
    s=60,
    marker="x",
    label="Test data"
)


# Plot all models
for name, model in models.items():

    y_plot = model.predict(x_plot)

    plt.plot(
        x_plot,
        y_plot,
        linewidth=2,
        label=name
    )


plt.xlabel("x")
plt.ylabel("y")

plt.title(
    "Degree-15 Polynomial Regression: "
    "No Regularization vs L1 vs L2 vs Elastic Net"
)

plt.legend()
plt.grid()

plt.ylim(-5, 10)

plt.show()


# =========================================================
# 8. Compare coefficients
# =========================================================

polynomial = models["No Regularization"].named_steps[
    "polynomialfeatures"
]

feature_names = polynomial.get_feature_names_out(["x"])

coefficients = {}

coefficients["No Regularization"] = (
    models["No Regularization"]
    .named_steps["linearregression"]
    .coef_
)

coefficients["L1 (Lasso)"] = (
    models["L1 (Lasso)"]
    .named_steps["lasso"]
    .coef_
)

coefficients["L2 (Ridge)"] = (
    models["L2 (Ridge)"]
    .named_steps["ridge"]
    .coef_
)

coefficients["Elastic Net"] = (
    models["Elastic Net"]
    .named_steps["elasticnet"]
    .coef_
)


plt.figure(figsize=(15, 8))

x_coeff = np.arange(len(feature_names))

for name, coef in coefficients.items():

    plt.plot(
        x_coeff,
        coef,
        marker="o",
        label=name
    )


plt.xticks(
    x_coeff,
    feature_names,
    rotation=45
)

plt.xlabel("Polynomial feature")
plt.ylabel("Coefficient value")

plt.title(
    "Polynomial Coefficients Under Different Regularization Methods"
)

plt.legend()
plt.grid()

plt.show()
```
