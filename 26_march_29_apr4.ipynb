```python
# Utilizing the Matrix Multiplication for calculating the cosine similarities between people
```

```python
import numpy as np
M = np.array([ # Three people and four movies rating 
    [.1, .2, .4],
    [.2, .2, .3],
    [.3, .2, .1],
    [.1, .2, .1],
])
```

```python
sqM = np.square(M)
print(sqM)
norms = np.sqrt(sqM.sum(axis=0))
norms
```

```python
uM = M / norms
```

```python
uM
```

```python
umT = uM.T
umT
```

```python
umT @ uM
```

```python
# Exercise 

def dy_dx(f, x):
    eps = 1e-8 # .00000001
    y1 = f(x)
    y2 = f(x+eps)
    return (y2-y1)/eps
```

```python
def f(x):
    return 10

dy_dx(f, 10)
```

```python
def f(x):
    return x

dy_dx(f, 10)
```

```python
def f(x):
    return x**2

print(dy_dx(f, 10)) # 2 x
print(dy_dx(f, 20))
```

```python
def f1(x):
    return x**2

def f2(y1):
    return y1**2

x = 10
y1 = f1(x)
y = f2(y1)

r1 = dy_dx(f1, x)
print(f"r1: {r1}")
r2 = dy_dx(f2, y1)
print(f"r2: {r2}")
print("r1*r2: ", r1*r2)
```
