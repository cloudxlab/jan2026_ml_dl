```python
# Softmax
import numpy as np


y_true = np.array([0, 0, 1])
output = np.array([1, 1, 5])

# Softmax
arr = np.exp(output)
output_softmax = arr / np.sum(arr)

print(output_softmax)

# Cross Entropy
result = y_true * np.log(output_softmax)
loss = -np.sum(result)
print(loss)
```
